```python
# Q30. Generate a report using print
import json

with open("employees.json", "r") as file:
    employees = json.load(file)
total_employees = len(employees)
missing_email = 0
missing_phone = 0
total_salary = 0
salary_count = 0
highest_salary = 0
lowest_salary = float ("inf")

for emp_id, details in employees.items():
    if details.get("Email") is None:
        missing_email += 1

    if details.get("Phone") is None:
            missing_phone += 1

    salary = details.get("Salary")
    if salary is not None:
        total_salary += salary
        salary_count += 1
    if salary > highest_salary:
        highest_salary = salary
    if salary < lowest_salary:
        lowest_salary = salary

if salary_count > 0:
    average_salary = total_salary / salary_count
else:
    average_salary = 0
    highest_salary = 0
    lowest_salary = 0

print("Total Employees:", total_employees)
print("Employees with Missing Email:", missing_email)
print("Employees with Missing Phone:", missing_phone)
print("Average Salary:", average_salary)
print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)

# Q31. Generate a report listing all employees department-wise
department_wise = {}

for emp_id, details in employees.items():
    dept = details.get("Department")
    name = details.get("Name")

    if dept not in department_wise:
        department_wise[dept] = []

    department_wise[dept].append(name)

for dept in sorted(department_wise):
    print("\n" + dept)
    print("-" * len(dept))
    for name in department_wise[dept]:
        print(name)


# Q32. Generate a report showing employees sorted by salary (highest to lowest)
sorted_employees = sorted(
employees.items(),
key=lambda item: item[1].get("Salary", 0),
reverse=True
)

print("\\nQ32. Employees sorted by salary (highest to lowest):")
for emp_id, details in sorted_employees:
    print(emp_id, details.get("Name"), details.get("Salary"))


# Q33. Find all employees whose names contain the word "25"
print('\\nQ33. Employees whose names contain "25":')
for emp_id, details in employees.items():
    name = details.get("Name")
    if name is not None and "25" in name:
        print(emp_id, name)


# Q34. Find all employees whose salary lies between ₹60,000 and ₹80,000
print("\\nQ34. Employees with salary between ₹60,000 and ₹80,000:")
for emp_id, details in employees.items():
    salary = details.get("Salary")
    if salary is not None and 60000 <= salary <= 80000:
        print(emp_id, details.get("Name"), salary)


# Q35. Find managers supervising more than 8 employees
manager_count = {}

for emp_id, details in employees.items():
    manager = details.get("Manager")
    if manager in manager_count:
        manager_count[manager] += 1
    else:
        manager_count[manager] = 1

print("\\nQ35. Managers supervising more than 8 employees:")
for manager, count in manager_count.items():
    if count > 8:
        print(manager, count)


# Q36. Find the department with the highest average salary
department_salary = {}
department_count = {}

for emp_id, details in employees.items():
    dept = details.get("Department")
    salary = details.get("Salary")

    if salary is not None:
        if dept in department_salary:
            department_salary[dept] += salary
            department_count[dept] += 1
        else:
            department_salary[dept] = salary
            department_count[dept] = 1

highest_avg_department = None
highest_avg_salary = 0

for dept in department_salary:
    avg_salary = department_salary[dept] / department_count[dept]
    if avg_salary > highest_avg_salary:
        highest_avg_salary = avg_salary
        highest_avg_department = dept

print("\\nQ36. Department with the highest average salary:")
print(highest_avg_department, highest_avg_salary)


# Q37. Identify employees with both email and phone number missing
print("\\nQ37. Employees with both email and phone missing:")
for emp_id, details in employees.items():
    if details.get("Email") is None and details.get("Phone") is None:
        print(emp_id, details.get("Name"))


# Q38. Find duplicate phone numbers
phone_count = {}

for emp_id, details in employees.items():
    phone = details.get("Phone")
    if phone is not None:
        if phone in phone_count:
            phone_count[phone] += 1
        else:
            phone_count[phone] = 1

print("\\nQ38. Duplicate phone numbers:")
for phone, count in phone_count.items():
    if count > 1:
        print(phone, "appears", count, "times")


# Q39. Find employees whose email domain is company.com
print("\\nQ39. Employees whose email domain is company.com:")
for emp_id, details in employees.items():
    email = details.get("Email")
    if email is not None and email.endswith("@company.com"):
        print(emp_id, details.get("Name"), email)


# Q40. Create a new dictionary containing only employees from the IT department
it_employees = {}

for emp_id, details in employees.items():
    if details.get("Department") == "IT":
        it_employees[emp_id] = details

print("\\nQ40. IT department employees dictionary:")
print(it_employees)
```
